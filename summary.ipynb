```python
from langchain_text_splitters  import CharacterTextSplitter
text_splitter = CharacterTextSplitter(
    separator="第[一二三四五六七八九十]+条|本合同附件为：|补充附页\n|\n\n|\n",
    is_separator_regex=True,
    chunk_size=200,
    chunk_overlap=20,
    length_function=len,
    keep_separator=True,
)

for i, s in enumerate(contract.text):
    if s[:4] == "补充附页":
        break

paragraphs = text_splitter.create_documents(['\n'.join(contract.text[:i])])

print(len(paragraphs))
paragraphs
```
