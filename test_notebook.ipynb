# search chain - rule parallel

````python
rule_list = []

for rule in list(rulers_manager.get_rulers(rulers_manager.rules_class[0]).rulers.values())[:]:
    rule_dict = {}
    rule_dict['rule_name']  = rule.rule_name
    rule_dict['tips']       = rule.tips
    rule_dict['elements']   = rule.elements
    rule_dict['rule']       = rule.reg_str
    rule_list.append(rule_dict)

#sentences = chain_search_sentence.batch(rule_list)
rule_list
````

````python
similarity_threshold = 1  # 定义相似度阈值
#total_rules = len(rule_list)
#matching_rules = 0

for rule_dict in rule_list:
    rule_embedding = get_text_embedding(rule_dict['tips'])
    D, I = index.search(np.array([rule_embedding]), 1)  # 查询最相似的三个向量
    success = False
    for dist, idx in zip(D[0], I[0]):
        most_similar_paragraph = paragraphs[idx]
        if dist < similarity_threshold:  # 距离阈值（越小越相似）
            rule_dict['search_from'] = most_similar_paragraph
            rule_dict['similarity_score'] = dist
            success = True
            break
    if not success:
        rule_dict['search_from'] = None
        rule_dict['similarity_score'] = None
````

````python
from operator import itemgetter
from langchain_core.runnables import RunnableLambda
from langchain_core.runnables import RunnableParallel

from langchain.globals import set_debug

set_debug(False)

chain_search_all_paragraphs = {}
for i, p in enumerate(paragraphs[:]):
    search_from_prompt = PromptTemplate(input_variables=[], template="现有待进行句子提取的合同文本如下:'''"+p+"'''")
    chain_search_all_paragraphs["idx" + str(i)] = (RunnableParallel(
                                                # search_from = search_from_prompt, 
                                               elements = itemgetter("elements"), 
                                               rule_name = itemgetter("rule_name"), 
                                               tips = itemgetter("tips"))
                                               | search_from_prompt + search_sentence_prompt 
                                               | Qwen2 
                                               | StrOutputParser())

def search_from_parser(input: list):
    import re
    pattern = r"'(.*?)'"
    match = re.search(pattern, input)

    if match:
        return match.group(1)
    else:
        return '无'  # 或者抛出异常，根据实际需求决定
    
chain_search_sentence = (  chain_search_all_paragraphs 
                         | RunnableLambda(lambda x: list(filter(lambda x: x != '' and x != '无' ,[search_from_parser(v) for _, v in x.items()])))
                        #  | RunnableLambda(search_from_parser)
                         | RunnableLambda(lambda x: "\n".join(f"- {item}" for item in x)))
````
